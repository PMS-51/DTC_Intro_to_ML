```python
median_hp=df['horsepower'].median()
mode_hp=df['horsepower'].mode()[0];
df['horsepower']=df['horsepower'].fillna(mode_hp)
new_median_hp=df['horsepower'].median();
if(median_hp>new_median_hp):
  print("Yes it decreased")
elif(median_hp<new_median_hp):
  print("Yes it increased")
else : print("No")
```

```python
df_asia=df[df['origin']=='Asia'][['vehicle_weight', 'model_year']]
```

```python
df_asia7 = df_asia.head(7)
display(df_asia7)
```

```python

X = df_asia7.values
print(f"Shape of X: {X.shape}")
```

```python

XTX = X.T @ X
print("XTX matrix:")
print(XTX)
```

```python

XTX_inv = np.linalg.inv(XTX)
print("Inverse of XTX:")
print(XTX_inv)
```

```python

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
print(f"Array y: {y}")
```

```python

w = XTX_inv @ X.T @ y
print("Resulting vector w:")
print(w)
```

```python

sum_w = w.sum()
print(f"Sum of all elements of w: {sum_w}")
```
